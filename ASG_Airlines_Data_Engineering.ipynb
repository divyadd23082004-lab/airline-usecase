```python
import pandas as pd
import numpy as np
```

```python
#Load all sheets from the Excel file

file_path = "/content/UseCase - Airlines.xlsx"

flights = pd.read_excel(file_path, sheet_name="flights")
bookings = pd.read_excel(file_path, sheet_name="bookings")
payments = pd.read_excel(file_path, sheet_name="payments")
passengers = pd.read_excel(file_path, sheet_name="passengers")

print("Data loaded successfully!")
print()
print("Flights:", flights.shape)
print("Bookings:", bookings.shape)
print("Payments:", payments.shape)
print("Passengers:", passengers.shape)
```

```python
#Display column names

print("FLIGHTS COLUMNS:")
print(flights.columns.tolist())
print("\n----- FLIGHTS -----")
display(flights.head())

print("\nBOOKINGS COLUMNS:")
print(bookings.columns.tolist())
print("\n----- BOOKINGS -----")
display(bookings.head())

print("\nPAYMENTS COLUMNS:")
print(payments.columns.tolist())
print("\n----- PAYMENTS -----")
display(payments.head())

print("\nPASSENGERS COLUMNS:")
print(passengers.columns.tolist())
print("\n----- PASSENGERS -----")
display(passengers.head())
```

```python
#Check data types and basic information

print("\n---- FLIGHTS ----")
flights.info()

print("\n---- BOOKINGS -----")
bookings.info()

print("\n---- PAYMENTS ----")
payments.info()

print("\n---- PASSENGERS ----")
passengers.info()
```

```python
#Check missing values

print("Missing values in FLIGHTS:")
print(flights.isnull().sum())

print("\nMissing values in BOOKINGS:")
print(bookings.isnull().sum())

print("\nMissing values in PAYMENTS:")
print(payments.isnull().sum())

print("\nMissing values in PASSENGERS:")
print(passengers.isnull().sum())
```

```python
#Check exact duplicate rows

print("Duplicate rows in FLIGHTS:", flights.duplicated().sum())
print("Duplicate rows in BOOKINGS:", bookings.duplicated().sum())
print("Duplicate rows in PAYMENTS:", payments.duplicated().sum())
print("Duplicate rows in PASSENGERS:", passengers.duplicated().sum())
```

```python
#Check important categorical values

print("Airlines:")
print(flights["airline"].value_counts(dropna=False))

print("\nBooking Status:")
print(bookings["status"].value_counts(dropna=False))

print("\nPayment Methods:")
print(payments["payment_method"].value_counts(dropna=False))
```

```python
#Check duplicate IDs

print("Duplicate flight IDs:",
      flights["flight_id"].duplicated().sum())
print("\n ---- DUPLICATE FLIGHT IDs ----")
print(flights[flights["flight_id"].duplicated(keep=False)].sort_values("flight_id"))

print("Duplicate booking IDs:",
      bookings["booking_id"].duplicated().sum())

print("Duplicate passenger IDs:",
      passengers["passenger_id"].duplicated().sum())

print("Duplicate payment IDs:",
      payments["payment_id"].duplicated().sum())
print("\n---- DUPLICATE PASSENGER IDs -----")
print(passengers[passengers["passenger_id"].duplicated(keep=False)].sort_values("passenger_id"))
```

```python
#Check flight time columns

print(flights[["flight_id", "departure_time", "arrival_time", "duration"]].head(10))

print("\nData types:")
print(flights[["departure_time", "arrival_time", "duration"]].dtypes)
```

```python
#Convert flight time to datetime

flights["departure_time"] = pd.to_datetime(
    flights["departure_time"],
    errors="coerce")

flights["arrival_time"] = pd.to_datetime(
    flights["arrival_time"],
    errors="coerce")

print(flights[["departure_time", "arrival_time"]].dtypes)
```

```python
#Calculate duration from departure and arrival

flights["calculated_duration"] = (
    flights["arrival_time"] - flights["departure_time"])

flights["calculated_duration_hours"] = (
    flights["calculated_duration"].dt.total_seconds() / 3600)

display(
    flights[
        [
            "flight_id",
            "departure_time",
            "arrival_time",
            "calculated_duration_hours"
        ]].head(10))
```

```python
#Find negative durations

negative_duration = flights[
    flights["calculated_duration_hours"] < 0]

print("Number of negative durations:",
      len(negative_duration))

display(
    negative_duration[[
            "flight_id",
            "departure_time",
            "arrival_time",
            "calculated_duration_hours"]])
```

```python
#Compare existing duration with calculated duration

flights['duration'] = pd.to_timedelta(flights['duration'].astype(str), errors='coerce')

flights["existing_duration_hours"] = (
    flights["duration"].dt.total_seconds() / 3600)

flights["duration_difference"] = (
    flights["existing_duration_hours"]
    - flights["calculated_duration_hours"])

display(
    flights[[
            "flight_id",
            "existing_duration_hours",
            "calculated_duration_hours",
            "duration_difference"]].head(10))
```

```python
#Find duration mismatches

duration_mismatch = flights[
    flights["duration_difference"].abs() > 0.01]
print("Number of duration mismatches:", len(duration_mismatch))
display(duration_mismatch)
```

```python
# Check negative durations

negative_duration = flights[
    flights["calculated_duration_hours"] < 0]
print("Number of negative durations:", len(negative_duration))
display(
    negative_duration[[
            "flight_id",
            "departure_time",
            "arrival_time",
            "calculated_duration_hours",
            "existing_duration_hours"]])
```

```python
#Handle overnight flight

flights["corrected_arrival_time"] = flights["arrival_time"]
overnight = flights["arrival_time"] < flights["departure_time"]
flights.loc[overnight, "corrected_arrival_time"] = (
    flights.loc[overnight, "arrival_time"] + pd.Timedelta(days=1))

flights["corrected_duration_hours"] = (
    flights["corrected_arrival_time"]
    - flights["departure_time"]).dt.total_seconds() / 3600

display(
    flights[overnight][[
            "flight_id",
            "departure_time",
            "arrival_time",
            "corrected_arrival_time",
            "corrected_duration_hours"]])
```

```python
#Validate corrected duration

negative_after_correction = flights[
    flights["corrected_duration_hours"] < 0]

print("Negative durations after correction:",len(negative_after_correction))
display(
    negative_after_correction[[
            "flight_id",
            "departure_time",
            "arrival_time",
            "corrected_duration_hours"]])
```

```python
#Missing value analysis

datasets = {
    "Flights": flights,
    "Bookings": bookings,
    "Payments": payments,
    "Passengers": passengers}

for name, df in datasets.items():
    print(f"\n----{name}----")
    missing = df.isnull().sum()
    percentage = (missing / len(df)) * 100

    result = pd.DataFrame({
        "Missing Values": missing,
        "Missing Percentage": percentage.round(2)})

    display(result[result["Missing Values"] > 0])
```

```python
#Check booking status
print(bookings["status"].value_counts(dropna=False))
```

```python
#Check payment amounts

print(payments["amount"].dtype)
print("\nNon-numeric amounts:")
print(payments[
    pd.to_numeric(payments["amount"], errors="coerce").isna()]["amount"].value_counts(dropna=False))
```

```python
#Create numeric payment amount
payments["amount_numeric"] = pd.to_numeric(
    payments["amount"],
    errors="coerce")
display(payments[["amount", "amount_numeric"]].head(10))
```

```python
#Passenger ID duplicate analysis

duplicate_passenger_ids = passengers[passengers["passenger_id"].duplicated(keep=False)]
print("Rows having duplicate passenger IDs:",len(duplicate_passenger_ids))
display(duplicate_passenger_ids.sort_values("passenger_id").head(20))
```

```python
#Exact duplicate analysis

print("Flights:", flights.duplicated().sum())
print("Bookings:", bookings.duplicated().sum())
print("Payments:", payments.duplicated().sum())
print("Passengers:", passengers.duplicated().sum())
```

```python
#Validate booking flight IDs

print("\n----Check booking → flight----")
missing_flights = bookings[~bookings["flight_id"].isin(flights["flight_id"])]
print("Booking records with missing flight IDs:",len(missing_flights))


print("\n-----Check booking → passenger-----")
missing_passengers = bookings[~bookings["passenger_id"].isin(passengers["passenger_id"])]
print("Booking records with missing passenger IDs:",len(missing_passengers))


print("\n----Check payment → booking------")
missing_bookings = payments[~payments["booking_id"].isin(bookings["booking_id"])]
print("Payment records with missing booking IDs:",len(missing_bookings))
```

```python
#Identify PII columns

print("Passenger PII columns:")
print([
    "first_name",
    "last_name",
    "email",
    "phone",
    "aadhaar_id",
    "date_of_birth"])

print("\nBooking PII columns:")
print([
    "passport_number",
    "emergency_contact_name",
    "emergency_contact_phone"])
```

```python
#Check Aadhaar ID lengths

aadhaar_length = passengers["aadhaar_id"].astype(str).str.len()
print(aadhaar_length.value_counts().sort_index())
```

```python
#Data quality summary

quality_summary = pd.DataFrame({
    "Dataset": ["Flights", "Bookings", "Payments", "Passengers"],
    "Rows": [
        len(flights),
        len(bookings),
        len(payments),
        len(passengers)],
    "Columns": [
        flights.shape[1],
        bookings.shape[1],
        payments.shape[1],
        passengers.shape[1]],
    "Missing Values": [
        flights.isnull().sum().sum(),
        bookings.isnull().sum().sum(),
        payments.isnull().sum().sum(),
        passengers.isnull().sum().sum()],
    "Duplicate Rows": [
        flights.duplicated().sum(),
        bookings.duplicated().sum(),
        payments.duplicated().sum(),
        passengers.duplicated().sum()]})
display(quality_summary)
```

```python
#Remove exact duplicate rows

flights = flights.drop_duplicates().reset_index(drop=True)
bookings = bookings.drop_duplicates().reset_index(drop=True)
payments = payments.drop_duplicates().reset_index(drop=True)
passengers = passengers.drop_duplicates().reset_index(drop=True)

print("Duplicate rows after cleaning:")
print("Flights:", flights.duplicated().sum())
print("Bookings:", bookings.duplicated().sum())
print("Payments:", payments.duplicated().sum())
print("Passengers:", passengers.duplicated().sum())
```

```python
#Clean airline values

flights["airline"] = flights["airline"].fillna("Unknown")
flights["airline"] = flights["airline"].replace(
    "UNKNOWN", "Unknown")
print(flights["airline"].value_counts())
```

```python
# Step 33: Clean booking status

valid_status = ["CONFIRMED", "CANCELLED", "PENDING"]
bookings["status"] = bookings["status"].where(
    bookings["status"].isin(valid_status),
    "UNKNOWN")
print(bookings["status"].value_counts())

#Check invalid payment amounts
print("\nInvalid or missing payment amounts:",payments["amount_numeric"].isna().sum())
```

```python
#Check valid payment amounts

print("Minimum payment amount:",payments["amount_numeric"].min())
print("Maximum payment amount:",payments["amount_numeric"].max())
print("Missing payment amounts:",payments["amount_numeric"].isna().sum())
```

```python
#Create final flight duration

flights["flight_duration_hours"] = (
    flights["corrected_duration_hours"])
print(flights[["flight_id", "flight_duration_hours"]].head(10))
```

```python
#Create flight analysis features

flights["route"] = (flights["source"] + " → " + flights["destination"])
flights["departure_date"] = (flights["departure_time"].dt.date)
flights["departure_hour"] = (flights["departure_time"].dt.hour)
flights["departure_day"] = (flights["departure_time"].dt.day_name())
flights["departure_month"] = (flights["departure_time"].dt.month_name())
flights["overnight_flag"] = (flights["arrival_time"] < flights["departure_time"])
display(
    flights[[
            "flight_id",
            "route",
            "flight_duration_hours",
            "departure_hour",
            "departure_day",
            "departure_month",
            "overnight_flag"]].head(10))
```

```python
#Validate final flight data

print("Total flights:", len(flights))
print("Negative final durations:",(flights["flight_duration_hours"] < 0).sum())
print("Missing final durations:",flights["flight_duration_hours"].isna().sum())
print("Missing routes:",flights["route"].isna().sum())
```

```python
#Passenger cleaning
# Remove only exact duplicate rows
passengers = passengers.drop_duplicates().reset_index(drop=True)
# Fill missing names
passengers["first_name"] = passengers["first_name"].fillna("Unknown")
passengers["last_name"] = passengers["last_name"].fillna("Unknown")
print("Passenger rows after removing exact duplicates:", len(passengers))
display(passengers.head())
```

```python
#Mask Passenger PII

passengers["email_masked"] = passengers["email"].astype(str).str.replace(
    r"(^.).*(@.*$)", r"\1***\2", regex=True)

passengers["phone_masked"] = passengers["phone"].astype(str).apply(
    lambda x: "***" + x[-4:] if x != "nan" else "Unknown")

passengers["aadhaar_masked"] = passengers["aadhaar_id"].astype(str).apply(
    lambda x: "********" + x[-4:] if x != "nan" else "Unknown")

display(passengers[
    ["passenger_id", "first_name", "last_name",
     "email_masked", "phone_masked", "aadhaar_masked"]].head())
```

```python

#Clean Booking Status

valid_status = ["CONFIRMED", "CANCELLED", "PENDING"]
bookings["status_clean"] = bookings["status"].where(bookings["status"].isin(valid_status),"UNKNOWN")
print(bookings["status_clean"].value_counts())
```

```python
#Convert payment amount to numeric

payments["amount_numeric"] = pd.to_numeric(payments["amount"],errors="coerce")
print("Missing/Invalid payment amounts:",payments["amount_numeric"].isna().sum())
display(payments.head())
```

```python
#Referential Integrity Validation

invalid_booking_flights = bookings[~bookings["flight_id"].isin(flights["flight_id"])]
invalid_booking_passengers = bookings[~bookings["passenger_id"].isin(passengers["passenger_id"])]
invalid_payment_bookings = payments[~payments["booking_id"].isin(bookings["booking_id"])]

print("Invalid booking-flight relationships:", len(invalid_booking_flights))
print("Invalid booking-passenger relationships:", len(invalid_booking_passengers))
print("Invalid payment-booking relationships:", len(invalid_payment_bookings))
```

```python
#Airline KPI

airline_kpi = flights.groupby("airline").agg(
    Total_Flights=("flight_id", "count"),
    Average_Duration_Hours=("flight_duration_hours", "mean")).reset_index()
airline_kpi["Average_Duration_Hours"] = (airline_kpi["Average_Duration_Hours"].round(2))
display(airline_kpi)
```

```python
#Route KPI

route_kpi = flights.groupby("route").agg(
    Total_Flights=("flight_id", "count"),
    Average_Duration_Hours=("flight_duration_hours", "mean")).reset_index()
route_kpi["Average_Duration_Hours"] = (
    route_kpi["Average_Duration_Hours"].round(2))
route_kpi = route_kpi.sort_values(
    "Total_Flights",
    ascending=False)
display(route_kpi.head(10))
```

```python
#Booking KPI

booking_kpi = bookings.groupby("status_clean").size().reset_index(name="Total_Bookings")
display(booking_kpi)

print("\n-----payment KPI----")
payment_kpi = payments.groupby("payment_method").agg(
    Total_Transactions=("payment_id", "count"),
    Total_Amount=("amount_numeric", "sum")).reset_index()
payment_kpi["Total_Amount"] = payment_kpi["Total_Amount"].round(2)
display(payment_kpi)
```

```python
#Create safe Passenger table for reporting

passengers_report = passengers[[
        "passenger_id",
        "first_name",
        "last_name",
        "age",
        "gender",
        "email_masked",
        "phone_masked",
        "aadhaar_masked"]].copy()
display(passengers_report.head())
```

```python
#Create safe Booking table for reporting

bookings_report = bookings[[
        "booking_id",
        "passenger_id",
        "flight_id",
        "booking_date",
        "status_clean",
        "seat_number"]].copy()
display(bookings_report.head())
```

```python
#Final Pipeline Validation

#Create safe Booking table
bookings_report = bookings[[
        "booking_id",
        "passenger_id",
        "flight_id",
        "booking_date",
        "status_clean",
        "seat_number"]].copy()

#Create safe Passenger table
passengers_report = passengers[[
        "passenger_id",
        "first_name",
        "last_name",
        "age",
        "gender",
        "email_masked",
        "phone_masked",
        "aadhaar_masked"]].copy()

print("------ FINAL DATA VALIDATION ------")

print("Flights:", len(flights))
print("Bookings:", len(bookings_report))
print("Payments:", len(payments))
print("Passengers:", len(passengers_report))

print("\nNegative flight durations:",
      (flights["flight_duration_hours"] < 0).sum())

print("Missing flight durations:",
      flights["flight_duration_hours"].isna().sum())

print("Missing numeric payments:",
      payments["amount_numeric"].isna().sum())

print("\nPipeline validation completed.")
```
